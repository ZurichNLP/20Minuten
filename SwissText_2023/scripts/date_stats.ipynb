```python
import json
from pathlib import Path
from datetime import datetime
from tqdm import tqdm
```

```python
corpus_dir = 'resources/data/20_min_0_00_001/'
splits = ['20min_all_test', '20min_all_dev', '20min_all_train']

dates = {'year': [], 'year_month': []}

c = 0
for split in splits:
    files = (Path(corpus_dir) / f'{split}').glob('*.json')
    for file in tqdm(files):
        c += 1
        with open(file, 'r') as f:
            data = json.load(f)
            date = data['datePublished'][:10]
            y, m, d = date.split('-')
            dates['year'].append(int(y))
            dates['year_month'].append(f'{y}-{m}')
            
print(f'Number of articles: {c}')

```

```python
import pandas as pd
df = pd.DataFrame(dates)
```
